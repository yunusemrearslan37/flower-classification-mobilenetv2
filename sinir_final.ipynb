# Flower Image Classification with MobileNetV2

Kaggle [Flowers](https://www.kaggle.com/datasets/l3llff/flowers) veri setinde 16 çiçek sınıfını sınıflandıran transfer learning + fine-tuning notebook'u.

**Model:** MobileNetV2 (ImageNet ağırlıkları) · **Giriş:** 160×160 · **Metrikler:** Accuracy, Precision, Recall, F1, confusion matrix

**Test sonuçları:** Accuracy **91.52%** · Macro F1 **91.40%**

Google Colab'da GPU (T4) ile çalıştırın: Runtime → Change runtime type → GPU.

```python
# ============================================================
# FLOWERS (l3llff/flowers) - Hızlı Transfer Learning + Fine-tune
# Daha hızlı: IMG_SIZE=160, hafif augmentation, kısa epoch
# Metrikler: Accuracy, Precision, Recall, F1, Confusion Matrix
# + GRAFİKLER: Train/Val Accuracy ve Train/Val Loss
# ============================================================

!pip -q install kagglehub scikit-learn

import os, numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix, classification_report
import kagglehub

print("TF:", tf.__version__)
print("GPU:", tf.config.list_physical_devices('GPU'))

# -----------------------------
# 1) Dataset indir
# -----------------------------
path = kagglehub.dataset_download("l3llff/flowers")
data_dir = os.path.join(path, "flowers")
if not os.path.exists(data_dir):
    data_dir = path
print("data_dir:", data_dir)

# -----------------------------
# 2) Dataset load + split
# -----------------------------
IMG_SIZE = (160, 160)      # 224 yerine 160 -> ciddi hız
BATCH_SIZE = 32
SEED = 42

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="int"
)

temp_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=True,
    label_mode="int"
)

class_names = train_ds.class_names
num_classes = len(class_names)
print("Sınıf sayısı:", num_classes)
print("Sınıflar:", class_names)

# temp -> val + test
temp_card = tf.data.experimental.cardinality(temp_ds).numpy()
val_batches = temp_card // 2
val_ds  = temp_ds.take(val_batches)
test_ds = temp_ds.skip(val_batches)

AUTOTUNE = tf.data.AUTOTUNE

# NOT: cache RAM yetmezse yavaşlatabilir. İstersen cache() satırlarını kaldır.
train_ds = train_ds.shuffle(1000, seed=SEED).prefetch(AUTOTUNE)
val_ds   = val_ds.prefetch(AUTOTUNE)
test_ds  = test_ds.prefetch(AUTOTUNE)

# -----------------------------
# 3) Class weight (opsiyonel ama faydalı)
# -----------------------------
counts = np.zeros(num_classes, dtype=np.int64)
for _, y in train_ds:
    y = y.numpy()
    for i in range(num_classes):
        counts[i] += np.sum(y == i)
total = counts.sum()
class_weight = {i: (total / (num_classes * counts[i])) for i in range(num_classes) if counts[i] > 0}
print("Örnek sınıf dağılımı:", {class_names[i]: int(counts[i]) for i in range(num_classes)})

# -----------------------------
# 4) Augmentation (hafifletildi)
# -----------------------------
aug = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.15),
    layers.RandomZoom(0.15),
], name="aug")

# -----------------------------
# 5) Model: MobileNetV2 (daha hızlı) + Fine-tuning
# EfficientNet yerine MobileNetV2 seçtim -> hız artar, iyi performans verir.
# -----------------------------
preprocess = tf.keras.applications.mobilenet_v2.preprocess_input

inputs = keras.Input(shape=IMG_SIZE + (3,))
x = aug(inputs)
x = layers.Lambda(preprocess)(x)

base = tf.keras.applications.MobileNetV2(
    include_top=False,
    weights="imagenet",
    input_shape=IMG_SIZE + (3,)
)
base.trainable = False

x = base(x, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(num_classes, activation="softmax")(x)

model = keras.Model(inputs, outputs)
model.summary()

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=3e-4),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cbs = [
    keras.callbacks.EarlyStopping(monitor="val_accuracy", patience=4, restore_best_weights=True),
    keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.3, patience=2, min_lr=1e-6, verbose=1),
]

# -----------------------------
# 6) Stage-1: Head eğitimi (kısa)
# -----------------------------
EPOCHS1 = 8
h1 = model.fit(train_ds, validation_data=val_ds, epochs=EPOCHS1,
               class_weight=class_weight, callbacks=cbs, verbose=1)

# -----------------------------
# 7) Stage-2: Fine-tune (son 20 katman)
# -----------------------------
base.trainable = True
for layer in base.layers[:-20]:
    layer.trainable = False

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=1e-5),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

EPOCHS2 = 10
h2 = model.fit(train_ds, validation_data=val_ds,
               epochs=EPOCHS1 + EPOCHS2,
               initial_epoch=len(h1.history["loss"]),
               class_weight=class_weight, callbacks=cbs, verbose=1)

# -----------------------------
# 7.5) EĞİTİM GRAFİKLERİ (EKLENDİ)
# -----------------------------
# h1 ve h2 history birleştir
train_acc = h1.history.get("accuracy", []) + h2.history.get("accuracy", [])
val_acc   = h1.history.get("val_accuracy", []) + h2.history.get("val_accuracy", [])
train_loss = h1.history.get("loss", []) + h2.history.get("loss", [])
val_loss   = h1.history.get("val_loss", []) + h2.history.get("val_loss", [])

epochs_range = range(len(train_acc))

plt.figure(figsize=(18,5))

# Accuracy grafiği
plt.subplot(1,2,1)
plt.plot(epochs_range, train_acc, marker="o", label="Train Acc")
plt.plot(epochs_range, val_acc, marker="s", label="Val Acc")
plt.title("Eğitim / Doğrulama Doğruluk Grafiği")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.grid(True, alpha=0.3)
plt.legend()

# Loss grafiği
plt.subplot(1,2,2)
plt.plot(epochs_range, train_loss, marker="o", label="Train Loss")
plt.plot(epochs_range, val_loss, marker="s", label="Val Loss")
plt.title("Eğitim / Doğrulama Kayıp Grafiği")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()

# -----------------------------
# 8) Test metrikleri
# -----------------------------
y_true, y_pred = [], []
for images, labels in test_ds:
    probs = model.predict(images, verbose=0)
    y_true.extend(labels.numpy().tolist())
    y_pred.extend(np.argmax(probs, axis=1).tolist())

y_true = np.array(y_true); y_pred = np.array(y_pred)

acc = accuracy_score(y_true, y_pred)
p_macro, r_macro, f1_macro, _ = precision_recall_fscore_support(y_true, y_pred, average="macro", zero_division=0)

print("\n" + "="*60)
print("TEST SONUÇLARI")
print("="*60)
print(f"Accuracy: {acc:.4f}")
print(f"Precision (macro): {p_macro:.4f}")
print(f"Recall (macro):    {r_macro:.4f}")
print(f"F1-Score (macro):  {f1_macro:.4f}")

print("\nSınıf bazlı rapor:")
print(classification_report(y_true, y_pred, target_names=class_names, digits=4, zero_division=0))

cm = confusion_matrix(y_true, y_pred)
plt.figure(figsize=(10,8))
plt.imshow(cm, interpolation="nearest")
plt.title("Confusion Matrix (Test)")
plt.colorbar()
ticks = np.arange(num_classes)
plt.xticks(ticks, class_names, rotation=45, ha="right")
plt.yticks(ticks, class_names)
plt.xlabel("Tahmin"); plt.ylabel("Gerçek")
plt.tight_layout()
plt.show()

model.save("flowers_mobilenetv2_finetuned.keras")
print("\nKaydedildi: flowers_mobilenetv2_finetuned.keras")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from PIL import Image

# ✅ class isimleri (senin dataset sırası)
class_names = [
    'astilbe', 'bellflower', 'black_eyed_susan', 'calendula',
    'california_poppy', 'carnation', 'common_daisy', 'coreopsis',
    'daffodil', 'dandelion', 'iris', 'magnolia',
    'rose', 'sunflower', 'tulip', 'water_lily'
]

# 0) model var mı?
try:
    _ = model
except NameError:
    raise NameError("❌ model yok. Önce eğitim hücresini çalıştır (model oluşsun).")

# 1) Fotoğraf yükle (HER UZANTI)
print("📌 Test görselini seç (jpg/png/webp/jfif/heic vb.)")
up = files.upload()
img_path = list(up.keys())[0]
print("✅ Yüklendi:", img_path)

# 2) Aç + RGB
img = Image.open(img_path).convert("RGB")

# 3) Model input size
h, w = model.input_shape[1], model.input_shape[2]
img_resized = img.resize((w, h))

# 4) Batch
x = np.array(img_resized, dtype=np.float32)
x = np.expand_dims(x, axis=0)

# 5) Predict
probs = model.predict(x, verbose=0)[0]
pred_idx = int(np.argmax(probs))
conf = float(np.max(probs))
pred_name = class_names[pred_idx] if pred_idx < len(class_names) else f"class_{pred_idx}"

# 6) Göster
plt.figure(figsize=(5,5))
plt.imshow(img_resized)
plt.axis("off")
plt.title(f"Tahmin: {pred_name} | Güven: {conf*100:.2f}%")
plt.show()

print("➡️ Tahmin:", pred_name)
print("➡️ Güven:", f"{conf*100:.2f}%")

# 7) Top-3
top3 = probs.argsort()[-3:][::-1]
print("\nTop-3 Tahminler:")
for i in top3:
    name = class_names[i] if i < len(class_names) else f"class_{i}"
    print(f"- {name}: {probs[i]*100:.2f}%")
```
